```python
import json

with open('/content/drive/MyDrive/students.json', 'r') as file:
    students = json.load(file)

print(students)
```

```python
import ipywidgets as widgets
from IPython.display import display

student_dropdown = widgets.Dropdown(
    options=[
        (s["first_name"] + " " + s["last_name"], i)
        for i, s in enumerate(students)
    ],
    description="Student:"
)

display(student_dropdown)
```

```python
first_name = widgets.Text(description="First Name:")
last_name = widgets.Text(description="Last Name:")
email = widgets.Text(description="Email:")
courses = widgets.Textarea(description="Courses:")
link = widgets.Text(description="Link:")

def show_student(change):
    student = students[student_dropdown.value]

    first_name.value = student["first_name"]
    last_name.value = student["last_name"]
    email.value = student["email"]
    courses.value = ", ".join(student["courses"])
    link.value = student["interesting_link"]

student_dropdown.observe(show_student, names="value")

show_student(None)

display(first_name, last_name, email, courses, link)
```

```python
favorite_show = widgets.Text(
    description="Favorite Show:"
)

update_button = widgets.Button(
    description="Update",
    button_style="success"
)

message = widgets.Output()

def update_student(button):
    index = student_dropdown.value

    students[index]["favorite_show"] = favorite_show.value

    with open('/content/drive/MyDrive/students.json', 'w') as file:
        json.dump(students, file, indent=4)

    with message:
        message.clear_output()
        print("Student updated successfully!")

update_button.on_click(update_student)

display(favorite_show, update_button, message)
```
