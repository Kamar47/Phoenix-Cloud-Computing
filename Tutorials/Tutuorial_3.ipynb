```python


from firebase import firebase

URL = 'https://lab-3-bbadc-default-rtdb.europe-west1.firebasedatabase.app/'
FBconn = firebase.FirebaseApplication(URL, None)

PATH = '/words'


def clean_word(word):
    """lowercase + remove punctuation (Firebase keys can't have . # $ [ ] /)"""
    word = word.lower()
    return ''.join(ch for ch in word if ch.isalnum())


def get_count(word):
    """returns how many times a word appears (0 if it's not in the DB)"""
    count = FBconn.get(PATH, word)
    return count if count else 0


def add_word(word):
    """option 1 - add a single word (if it exists -> count + 1)"""
    word = clean_word(word)
    if word == '':
        return
    new_count = get_count(word) + 1
    FBconn.put(PATH, word, new_count)


def add_text(text):
    """option 2 - split the text into words and add each one"""
    for word in text.split():
        add_word(word)


def update_word(word, new_count):
    """option 3 - set the count of a word to a new number"""
    word = clean_word(word)
    if get_count(word) == 0:
        print(f"'{word}' is not in the database")
        return
    FBconn.put(PATH, word, new_count)
    print(f"'{word}' updated to {new_count}")


def delete_word(word):
    """option 4 - delete a word"""
    word = clean_word(word)
    if get_count(word) == 0:
        print(f"'{word}' is not in the database")
        return
    FBconn.delete(PATH, word)
    print(f"'{word}' deleted")


def show_all():
    """option 5 - print all saved words like: hello :2, world:1"""
    words = FBconn.get(PATH, None)
    if not words:
        print("No words saved yet")
        return
    print(', '.join(f"{w}:{c}" for w, c in words.items()))


# ---------------- MENU ----------------
while True:
    print("\nWord Counter Menu:")
    print("1. Add a single word")
    print("2. Add a full text")
    print("3. Update a word's count")
    print("4. Delete a word")
    print("5. View all words")
    print("6. Exit")

    choice = input("\nEnter your choice (1-6): ")

    if choice == '1':
        add_word(input("Enter a word: "))
        print("Added!")
    elif choice == '2':
        add_text(input("Enter text: "))
        print("Text added!")
    elif choice == '3':
        w = input("Which word? ")
        try:
            update_word(w, int(input("New count: ")))
        except ValueError:
            print("Count must be a number")
    elif choice == '4':
        delete_word(input("Which word? "))
    elif choice == '5':
        show_all()
    elif choice == '6':
        print("Bye!")
        break
    else:
        print("Invalid choice, try again")
```
